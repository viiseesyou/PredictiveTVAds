# Обучение бейзлайна

## Обучение простых моделей

### Линейная регрессия

```python
# Масштабирование признаков
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Обучение модели с замером времени
start_time = time.time()
lr_model = LinearRegression()
lr_model.fit(X_train_scaled, y_train)
end_time = time.time()

# Предсказания и метрика
lr_preds = lr_model.predict(X_test_scaled)
lr_r2 = r2_score(y_test, lr_preds)

print(f"[LinearRegression] Обучение завершено за {end_time - start_time:.2f} секунд.")
print(f"[LinearRegression] R² на тестовой выборке: {lr_r2:.5f}")
```

```python
from sklearn.metrics import mean_absolute_percentage_error

# Предсказания на тестовой выборке
lr_preds = lr_model.predict(X_test)

# Вычисление MAPE
lr_mape = mean_absolute_percentage_error(y_test, lr_preds) * 100  # в процентах

print(f"MAPE для линейной регрессии: {lr_mape:.2f}%")
```

### Случайный лес

```python
start_time = time.time()

rf_limited = RandomForestRegressor(
    n_jobs=-1,
    random_state=42,
    max_depth=15,
)

rf_limited.fit(X_train, y_train)

y_pred = rf_limited.predict(X_test)
r2 = r2_score(y_test, y_pred)

end_time = time.time()

print(f"[RF (ограничен)] Обучено за {end_time - start_time:.2f} секунд")
print(f"[RF (ограничен)] R² на test_dec: {r2:.4f}")
```

```python
rf_preds = rf_limited.predict(X_test)

rf_mape = mean_absolute_percentage_error(y_test, rf_preds) * 100  # в процентах
print(f"MAPE для Random Forest после удаления нулей: {rf_mape:.2f}%")
```
